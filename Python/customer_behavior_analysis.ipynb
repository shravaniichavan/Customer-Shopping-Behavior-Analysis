```python
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=20, kde=True)
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
df["Gender"].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Gender")
plt.title("Customers by Gender")
plt.show()
```

```python
category_sales = df.groupby("Category")["Purchase Amount (USD)"].sum().sort_values(ascending=False)
category_sales
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales (USD)")
plt.xticks(rotation=30)
plt.show()
```

```python
df["Item Purchased"].value_counts().head(10)
```

```python
top_items = df["Item Purchased"].value_counts().head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=top_items.values, y=top_items.index)
plt.title("Top 10 Most Purchased Items")
plt.xlabel("Number of Purchases")
plt.ylabel("Item")
plt.show()
```

```python
season_sales = df.groupby("Season")["Purchase Amount (USD)"].sum().sort_values(ascending=False)
season_sales
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x=season_sales.index, y=season_sales.values)
plt.title("Sales by Season")
plt.xlabel("Season")
plt.ylabel("Total Sales (USD)")
plt.show()
```

```python
df.groupby("Gender")["Purchase Amount (USD)"].mean().sort_values(ascending=False)
```

```python
total_revenue = df["Purchase Amount (USD)"].sum()
avg_purchase = df["Purchase Amount (USD)"].mean()
total_customers = df["Customer ID"].nunique()
avg_rating = df["Review Rating"].mean()

print("Total Revenue:", total_revenue)
print("Average Purchase:", avg_purchase)
print("Total Customers:", total_customers)
print("Average Rating:", avg_rating)
```

```python
df["Spending Category"] = pd.cut(
    df["Purchase Amount (USD)"],
    bins=[0, 50, 100, float("inf")],
    labels=["Low", "Medium", "High"]
)

df["Spending Category"].value_counts()
```
